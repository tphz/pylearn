## 图像测试文件及图像信息

````python
from PIL import Image
im = Image.open("catgang.jpg")
print(f"{im.mode=}, {im.format=}, {im.size=}")
print(f"{im.getpixel((5,3))=}")
im
````

````python
glass = Image.open("glass.png")
print(f"{glass.mode=}, {glass.format=}, {glass.size=}")
print(f"{glass.getpixel((5,3))=}")
glass
````

## 图像裁剪```Image.crop()```

````python
cat = im.crop((410, 530, 710, 830))  # 裁剪猫老大的头像
cat.save("catgang1.jpg", "JPEG")  # 可选的，保存输出为一个中间文件
print(f"{cat.mode=}, {cat.format=}, {cat.size=}")
print(f"{cat.getpixel((5,3))=}")
cat
````

## 图像调整大小```Image.resize()```

````python
# 将墨镜大小缩小为200，200；resize生成新Image对象
glass = glass.resize((200, 200)) 
# 注意mode不变，但format置为None了
print(f"{glass.mode=}, {glass.format=}, {glass.size=}")
print(f"{glass.getpixel((5,3))=}")
glass
````

## 图像粘贴```Image.paste()```

````python
# 贴到猫头像中，mask是遮罩
cat.paste(glass, (100, 100, 300, 300), mask=glass)
cat.save("catgang2.jpg")  # 保存为文件
print(f"{cat.mode=}, {cat.format=}, {cat.size=}")
print(f"{cat.getpixel((5,3))=}")
cat
````

## 图像缩略图```Image.thumbnail()```

````python
size = (128, 128)
cat.thumbnail(size)  # 缩略图版本，改变了cat图像对象（区别于resize）
cat.save("catgang1.gif", "GIF")  # 保存为GIF图像格式文件
print(f"{cat.mode=}, {cat.format=}, {cat.size=}")
print(f"{cat.getpixel((5,3))=}")
cat
````

## 完整代码

````python
from PIL import Image
im = Image.open("catgang.jpg")
cat = im.crop((410, 530, 710, 830))  # 裁剪猫老大的头像
glass = Image.open("glass.png")
# 将墨镜大小缩小为200，200；resize生成新Image对象
glass = glass.resize((200, 200))
# 贴到猫头像中，mask是遮罩
cat.paste(glass, (100, 100, 300, 300), mask=glass)  
cat.save("catgang2.jpg")  # 保存为文件
size = (128, 128)
cat.thumbnail(size)  # 缩略图版本，改变了cat图像对象（区别于resize）
cat.save("catgang1.gif", "GIF")  # 保存为GIF图像格式文件
````
