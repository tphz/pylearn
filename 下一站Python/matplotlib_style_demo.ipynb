
`matplotlib.pyplot.style` 是 matplotlib 的样式管理工具，它提供了一种简单的方式来自定义图表的外观。

````python
import matplotlib.pyplot as plt
import numpy as np

# 查看所有可用的样式
print("可用的样式列表：")
print(plt.style.available)
````


matplotlib 提供了多种预定义的样式，我们可以通过 `plt.style.use()` 来使用它们。下面我们用一些常用的样式来绘制相同的图表进行对比。

````python
# 创建示例数据
x = np.linspace(0, 10, 100)
y = np.sin(x)

# 创建一个函数来绘制相同的图表
def plot_with_style(style_name):
    with plt.style.context(style_name):
        plt.figure(figsize=(8, 4))
        plt.plot(x, y)
        plt.title(f"Style: {style_name}")
        plt.grid(True)
        plt.show()

# 展示几个常用样式
styles = ['default', 'seaborn-v0_8', 'ggplot', 'bmh', 'dark_background']
for style in styles:
    plot_with_style(style)
````


1. 全局设置样式：
```python
plt.style.use('seaborn')
```

2. 临时使用样式（使用上下文管理器）：
```python
with plt.style.context('seaborn'):
    # 绘图代码
```

3. 组合多个样式：
```python
plt.style.use(['seaborn', 'dark_background'])
```

````python
# 演示临时使用样式
plt.style.use('default')  # 重置为默认样式

# 设置中文字体
plt.rcParams.update({
    'font.sans-serif': ['Microsoft YaHei', 'SimHei', 'SimSun', 'Arial'],  # 按优先级设置字体
    'axes.unicode_minus': False,  # 解决负号显示问题
    'font.family': 'sans-serif'  # 使用无衬线字体
})

# 正常绘图
plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.title("默认样式（中文测试）")
plt.grid(True)
plt.show()

# 临时使用 seaborn 样式
with plt.style.context('seaborn-v0_8'):
    plt.figure(figsize=(8, 4))
    plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'SimSun', 'Arial']
    plt.rcParams['axes.unicode_minus'] = False
    plt.plot(x, y)
    plt.title("临时使用 seaborn 样式（中文测试）")
    plt.grid(True)
    plt.show()

# 再次绘图，会回到默认样式
plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.title("回到默认样式（中文测试）")
plt.grid(True)
plt.show()
````


除了使用预定义的样式，我们还可以通过 `plt.rcParams` 来自定义具体的样式参数。

````python
# 自定义样式参数
plt.style.use('default')  # 先重置样式
plt.rcParams.update({
    'font.size': 12,
    'font.family': 'Microsoft YaHei',  # 设置中文字体
    'axes.unicode_minus': False,  # 解决负号显示问题
    'lines.linewidth': 2,
    'lines.linestyle': '--',
    'grid.color': 'gray',
    'grid.linestyle': ':',
    'grid.alpha': 0.5,
    'axes.facecolor': '#f0f0f0'
})

plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.title("自定义样式参数（中文测试）")
plt.grid(True)
plt.show()
````


一些常用的样式参数：

1. 字体相关：
- `font.family`：字体家族
- `font.size`：字体大小
- `font.weight`：字体粗细

2. 线条相关：
- `lines.linewidth`：线宽
- `lines.linestyle`：线型
- `lines.color`：线条颜色

3. 网格相关：
- `grid.color`：网格颜色
- `grid.linestyle`：网格线型
- `grid.alpha`：网格透明度

4. 坐标轴相关：
- `axes.facecolor`：坐标区背景色
- `axes.grid`：是否显示网格
- `axes.titlesize`：标题字体大小

5. 图表相关：
- `figure.figsize`：图表大小
- `figure.dpi`：图表分辨率
- `figure.facecolor`：图表背景色
